```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
import pickle
```

```python
df = pd.read_csv("../data/text_dataset.csv")
df.head()
```

```python
genre_encoder = LabelEncoder()
y_genre = genre_encoder.fit_transform(df["genre"])

print("Unique genres:", list(genre_encoder.classes_))
print("First 10 encoded labels:", y_genre[:10])
```

```python
np.save("../models/y_genre.npy", y_genre)

with open("../models/genre_encoder.pkl", "wb") as f:
    pickle.dump(genre_encoder, f)

print("Genre labels and encoder saved")
```
