```python
import pandas as pd
import numpy as np
```

```python
# Load dataset
df = pd.read_csv("../data/text_dataset.csv")

# Display first 5 rows
df.head()
```

```python
# Rename column for convenience
df = df.rename(columns={"text_description": "text"})

df.head()
```

```python
# Convert text to lowercase
df['text'] = df['text'].str.lower()

# Remove leading/trailing spaces
df['text'] = df['text'].str.strip()

df.head()
```

```python
texts = df['text'].tolist()

print("Total samples:", len(texts))
print("Sample text:", texts[0])
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
# Create tokenizer
tokenizer = Tokenizer()

# Fit tokenizer on cleaned text
tokenizer.fit_on_texts(texts)

# Vocabulary size
vocab_size = len(tokenizer.word_index) + 1
print("Vocabulary size:", vocab_size)
```

```python
sequences = tokenizer.texts_to_sequences(texts)

print("Original text:", texts[0])
print("Tokenized sequence:", sequences[0])
```

```python
# Define maximum sentence length
max_len = 20

# Pad all sequences to same length
X_text = pad_sequences(sequences, maxlen=max_len, padding='post')

print("X_text shape:", X_text.shape)
print("First padded sequence:\n", X_text[0])
```

```python
import numpy as np
import pickle

# Save processed text data
np.save("../models/X_text.npy", X_text)

# Save tokenizer for later inference
with open("../models/text_tokenizer.pkl", "wb") as f:
    pickle.dump(tokenizer, f)

print("Text preprocessing completed and saved successfully")
```
