```python
import os
from music21 import converter, note, chord
from tqdm import tqdm
```

```python
midi_files = []

for root, dirs, files in os.walk("../data/midi_dataset"):
    for file in files:
        if file.lower().endswith((".mid", ".midi")):
            midi_files.append(os.path.join(root, file))

print("Total MIDI files found:", len(midi_files))
```

```python
all_notes = []
```

```python
MAX_FILES = 3000   # you can change to 2000 or 5000

all_notes = []
failed_files = 0

for file_path in tqdm(midi_files[:MAX_FILES]):
    try:
        midi = converter.parse(file_path)
        for element in midi.flat.notes:
            if isinstance(element, note.Note):
                all_notes.append(str(element.pitch))
            elif isinstance(element, chord.Chord):
                all_notes.append(".".join(str(n) for n in element.normalOrder))
    except Exception:
        failed_files += 1
        continue

print("Processed files:", MAX_FILES)
print("Total notes extracted:", len(all_notes))
print("Failed MIDI files:", failed_files)
```
