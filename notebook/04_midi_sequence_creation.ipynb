```python
import pickle

with open("../models/all_notes.pkl", "rb") as f:
    all_notes = pickle.load(f)

print("Total notes loaded:", len(all_notes))
```

```python
unique_notes = sorted(set(all_notes))

note_to_int = {note: i for i, note in enumerate(unique_notes)}
int_to_note = {i: note for note, i in note_to_int.items()}

print("Total unique notes:", len(unique_notes))
```

```python
sequence_length = 50

X_midi = []
y_midi = []

for i in range(len(all_notes) - sequence_length):
    seq_in = all_notes[i:i + sequence_length]
    seq_out = all_notes[i + sequence_length]

    X_midi.append([note_to_int[n] for n in seq_in])
    y_midi.append(note_to_int[seq_out])

print("Total sequences created:", len(X_midi))
```

```python
import numpy as np

X_midi = np.array(X_midi)
y_midi = np.array(y_midi)

# Normalize input (very important)
X_midi = X_midi / float(len(unique_notes))

print("X_midi shape:", X_midi.shape)
print("y_midi shape:", y_midi.shape)
```

```python
import pickle

np.save("../models/X_midi.npy", X_midi)
np.save("../models/y_midi.npy", y_midi)

with open("../models/note_to_int.pkl", "wb") as f:
    pickle.dump(note_to_int, f)

with open("../models/int_to_note.pkl", "wb") as f:
    pickle.dump(int_to_note, f)

print("MIDI sequence data saved successfully")
```
